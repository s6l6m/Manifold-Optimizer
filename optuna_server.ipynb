# Optuna Dashboard Server

### Imports

```python
import optuna_dashboard

from config import Config
from networking import check_port_in_use, get_url_for_port
```

### Load configuration

```python
CONFIG = Config(
    # Optuna Dashboard settings
    DASHBOARD_HOST="0.0.0.0",
    DASHBOARD_DOMAIN="localhost"
)
```

## Run Optuna Dashboard Server

```python
print(
    get_url_for_port(
        port=CONFIG.DASHBOARD_PORT,
        host=CONFIG.DASHBOARD_DOMAIN,
        protocol=CONFIG.DASHBOARD_PROTOCOL,
    )
)
if not check_port_in_use(CONFIG.DASHBOARD_PORT):
    optuna_dashboard.run_server(
        CONFIG.OPTUNA_SQLITE, host=CONFIG.DASHBOARD_HOST, port=CONFIG.DASHBOARD_PORT
    )
```
